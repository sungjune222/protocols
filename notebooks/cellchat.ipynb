#### Making a CellChat object from the h5ad file

```r
h5ad_file <- "nimlab_0421_cellchat.h5ad"
species <- "mouse"
group_key <- "cell"
data_dir <- "CELLCHAT"

invisible(
  suppressPackageStartupMessages({
    library(circlize)
    library(dotenv)
    library(rprojroot)
    library(CellChat)
    library(ComplexHeatmap)
  })
)

root <- rprojroot::find_root(rprojroot::has_file("pixi.toml"))
env_path <- file.path(root, ".env")
dotenv::load_dot_env(file = env_path)

r_dir <- file.path(root, "R")
r_scripts <- list.files(
  path = r_dir,
  pattern = "\\.R$",
  full.names = TRUE,
  recursive = TRUE
)
r_scripts <- r_scripts[!grepl("install\\.R$", r_scripts)]
invisible(sapply(r_scripts, function(x) try(source(x), silent = TRUE)))

cellchat <- run_cellchat(
  h5ad_file = h5ad_file,
  species = species,
  group_key = group_key,
  data_dir = data_dir
)

base_name <- tools::file_path_sans_ext(h5ad_file)
cellchat_dir <- get_env_dir("CELLCHAT")
cellchat_dir <- file.path(cellchat_dir, base_name)
```

```r
remove_groups <- c("cluster_PNSNeuron", "cluster_Schwann", "cluster_Osteoclast")
target_groups <- c("cluster_neutrophil_2")

group_order <- sort(levels(cellchat@idents))
group_order <- setdiff(group_order, remove_groups)

mat_count <- cellchat@net$count[group_order, group_order, drop = FALSE]
mat_weight <- cellchat@net$weight[group_order, group_order, drop = FALSE]

pair_count <- as.data.frame(as.table(mat_count), stringsAsFactors = FALSE)
colnames(pair_count) <- c("source", "target", "count")

pair_weight <- as.data.frame(as.table(mat_weight), stringsAsFactors = FALSE)
colnames(pair_weight) <- c("source", "target", "weight")

pair_directed <- merge(
  pair_count,
  pair_weight,
  by = c("source", "target")
)

if (length(target_groups) > 0) {
  pair_directed <- pair_directed[
    pair_directed$source %in% target_groups | pair_directed$target %in% target_groups, 
  ]
}

pair_directed <- pair_directed[
  pair_directed$count > 0 &
    pair_directed$source != pair_directed$target,
]

pair_directed <- pair_directed[
  order(pair_directed$count, pair_directed$weight, decreasing = TRUE),
]

pair_undirected <- transform(
  pair_directed,
  cell1 = pmin(source, target),
  cell2 = pmax(source, target)
)

pair_undirected <- aggregate(
  cbind(count, weight) ~ cell1 + cell2,
  data = pair_undirected,
  FUN = sum
)

pair_undirected <- pair_undirected[
  order(pair_undirected$weight, pair_undirected$count, decreasing = TRUE),
]

head(pair_undirected, 50)
```

```r
interest <- "COL4A2_ITGA3_ITGB1"
cellchat_lr <- subsetCommunication(cellchat, slot.name = "net")
cellchat_lr <- cellchat_lr[order(cellchat_lr$prob, decreasing = TRUE), ]

lr_interest <- cellchat_lr[cellchat_lr$interaction_name == interest, ]
lr_plot <- lr_interest[, c("source", "target", "prob")]

group_cell <- sort(levels(cellchat@idents))
grid_color <- scales::hue_pal()(length(group_cell))
names(grid_color) <- group_cell

min_prob <- min(lr_plot$prob)
max_prob <- max(lr_plot$prob)
col_fun <- colorRamp2(
  breaks = c(min_prob, (min_prob + max_prob)/2, max_prob),
  colors = c("#0A0F8F", "#FFEE58", "#FF0000")
)
link_colors <- col_fun(lr_plot$prob)

invisible({
  graphics.off()

  svglite::svglite(
    file.path(cellchat_dir, paste0(interest, "_lr.svg")),
    width = 9, height = 8
  )

  circos.clear()
  circos.par(
    cell.padding = c(0, 0, 0, 0),
    track.margin = c(0.01, 0.01),
    points.overflow.warning = FALSE,
    gap.after = 5
  )

  chordDiagram(
    x = lr_plot,
    order = group_cell,
    grid.col = grid_color,
    col = link_colors,
    transparency = 0.2,
    directional = 1,
    direction.type = c("diffHeight"),
    diffHeight = -0.03,
    link.sort = TRUE,
    link.largest.ontop = TRUE,
    annotationTrack = "grid",
    preAllocateTracks = list(track.height = 0.3)
  )

  circos.track(track.index = 1, panel.fun = function(x, y) {
    circlize::circos.text(
      CELL_META$xcenter,
      CELL_META$ylim[1],
      CELL_META$sector.index,

      facing = "clockwise",
      niceFacing = TRUE,
      adj = c(0, 0.5),
      cex = 0.8
    )
  }, bg.border = NA)

  lgd <- Legend(
    col_fun = col_fun,
    title = "Interaction\nStrength",
    at = c(min_prob, max_prob),
    labels = c("Low", "High")
  )
  draw(lgd, x = unit(0.9, "npc"), y = unit(0.9, "npc"))

  dev.off()
})
```

```r
cellchat_pathway <- subsetCommunication(cellchat, slot.name = "netP")
cellchat_pathway <- cellchat_pathway[cellchat_pathway$prob > 0, ]

cellchat_pathway$source <- as.character(cellchat_pathway$source)
cellchat_pathway$target <- as.character(cellchat_pathway$target)

pathway_count <- aggregate(
  pathway_name ~ source + target,
  data = cellchat_pathway,
  FUN = function(x) length(unique(x))
)

pathway_weight <- aggregate(
  prob ~ source + target,
  data = cellchat_pathway,
  FUN = sum
)

pair_pathway_directed <- merge(
  pathway_count,
  pathway_weight,
  by = c("source", "target")
)

colnames(pair_pathway_directed) <- c(
  "source", "target", "pathway_count", "weight"
)

pair_pathway_directed <- pair_pathway_directed[
  pair_pathway_directed$source != pair_pathway_directed$target,
]

pair_pathway_undirected <- transform(
  pair_pathway_directed,
  cell1 = pmin(source, target),
  cell2 = pmax(source, target)
)

pair_pathway_undirected <- aggregate(
  cbind(pathway_count, weight) ~ cell1 + cell2,
  data = pair_pathway_undirected,
  FUN = sum
)

pair_pathway_undirected <- pair_pathway_undirected[
  order(
    pair_pathway_undirected$pathway_count,
    pair_pathway_undirected$weight,
    decreasing = TRUE
  ),
]

head(pair_pathway_undirected, 30)
```

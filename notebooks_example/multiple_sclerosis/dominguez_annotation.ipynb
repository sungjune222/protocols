```python
assert isinstance(adata.obs, pd.DataFrame)
adata.obs["series"] = adata.obs["project_id"]
adata.obs["sample"] = adata.obs["sample"]
adata.obs["donor"] = adata.obs["sample"]
adata.obs["celltype"] = adata.obs["cell"]
adata.obs["cell"] = adata.obs["cell"]
adata.obs["leiden"] = adata.obs["leiden"]

keep = ["series", "sample", "donor", "celltype", "cell", "leiden"]
adata.obs.drop(columns = [c for c in adata.obs.columns if c not in keep], inplace=True) #type: ignore
adata.write_h5ad(os.path.join(final_h5ad_dir, "final.h5ad"))
```
